```python
import os, shutil, random

# Paths to your datasets
datasetA = r"D:\IITBHU Internship\other datasets\Drone1_F1\drone_dataset_yolo\dataset_txt"   # mixed folder (images + labels)
datasetB = r"D:\IITBHU Internship\other datasets\Drone3_F1\Database1"   # mixed folder (images + labels)
datasetC_images = r"D:\IITBHU Internship\other datasets\Drone2_F2\drone_dataset\train\images"
datasetC_labels = r"D:\IITBHU Internship\other datasets\Drone2_F2\drone_dataset\train\labels"

# Master dataset
base = "DroneDatasetCombined"
for sub in ["images/train","images/val","images/test",
            "labels/train","labels/val","labels/test"]:
    os.makedirs(os.path.join(base, sub), exist_ok=True)

def collect_pairs(img_dir, label_dir=None):
    pairs = []
    for f in os.listdir(img_dir):
        if f.lower().endswith((".jpg",".png",".jpeg")):
            img_path = os.path.join(img_dir, f)
            if label_dir:  # separate labels folder
                label_path = os.path.join(label_dir, os.path.splitext(f)[0] + ".txt")
            else:          # same folder
                label_path = os.path.join(img_dir, os.path.splitext(f)[0] + ".txt")
            if os.path.exists(label_path):
                pairs.append((img_path, label_path))
    return pairs

# Collect all image-label pairs
pairs = []
pairs += collect_pairs(datasetA)
pairs += collect_pairs(datasetB)
pairs += collect_pairs(datasetC_images, datasetC_labels)

# Shuffle
random.shuffle(pairs)

# Split
train_split = int(0.7 * len(pairs))
val_split = int(0.9 * len(pairs))

splits = {
    "train": pairs[:train_split],
    "val": pairs[train_split:val_split],
    "test": pairs[val_split:]
}

# Copy into unified structure
for split, items in splits.items():
    for img_path, label_path in items:
        fname = os.path.basename(img_path)
        lname = os.path.basename(label_path)

        shutil.copy(img_path, os.path.join(base,"images",split,fname))
        shutil.copy(label_path, os.path.join(base,"labels",split,lname))

print("✅ Combined dataset created at:", base)
```
